```python
import sympy as sp

f, f0, A, c, u0, D0 = sp.symbols('f f_0 A c u_0 D_0', positive=True)
fn = sp.symbols(r'\tilde{f}', positive=True)
```

```python
En = A / (1 + fn**c)
dEn = sp.diff(En, fn)
dEn
```

```python
J = u0 * f0**2 *fn**2 * En - D0 * f0**2 * fn**3 * dEn
sp.simplify(J)
```

```python
S = (1/f0) * sp.diff(J, fn)
S = sp.simplify(S)
S
```
